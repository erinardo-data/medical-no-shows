```python
df_sorted = df.sort_values(by=['Neighbourhood', 'Age', 'Gender'], ascending=[True, True, True], na_position='first')
df_sorted
```

Atomização da Sintaxe do Gerenciador de ContextoA palavra-chave with: Inicia a construção de um escopo isolado no Python. Ela invoca os métodos internos __enter__ (que altera temporariamente as configurações do Pandas) e __exit__ (que limpa tudo e restaura o padrão assim que a indentação acaba).A função display(): No Jupyter, se você apenas escrever o nome do objeto na última linha do bloco with, ele pode não renderizar de forma correta por limitações visuais de encadeamento do bloco. Usar a função nativa display(df_sorted) força o Jupyter a renderizar a tabela HTML interativa de forma explícita enquanto o contexto ainda está ativo.

```python
with pd.option_context('display.min_rows', 20, 'display.max_columns', 0):
    display(pd.get_option('display.min_rows'))
    display(pd.get_option('display.max_columns'))
    display(df_sorted)
```
